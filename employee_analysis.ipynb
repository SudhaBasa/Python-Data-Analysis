```python
import pandas as pd
```

```python
data = {
    'Employee': ['John', 'Mary', 'David'],
    'Salary': [50000, 60000, 55000]
}
```

```python
df = pd.DataFrame(data)
df
```

```python
df['Salary'].mean()
```

```python
# Calculate the minimum salary
minimum_salary = df['Salary'].min()
print('Minimum Salary:', minimum_salary)
```
